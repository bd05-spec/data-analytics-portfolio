# Ecommerce Multi-Category Store Analytics Pipeline

This notebook consolidates the end-to-end analytics workflow for 109.95M user behavior events (October - November 2019).
It covers chunked ETL streaming, schema validation, data profiling, visual generation, and automated cross-checking.

## Step 1: High-Performance Data Cleaning & Profiling

I'm setting up a chunked streaming pipeline (chunk size = 1,000,000 rows) to process the ~14.7 GB raw CSV data without running out of memory. This step validates schemas, enforces null handling rules, flags missing brand/category data as 'unknown', and derives price band features.

```python
# In execution, 01_clean_and_profile.py processes raw CSVs in 1M chunks.
# Below is the high-level configuration and streaming logic summary.
import pandas as pd
import numpy as np
from pathlib import Path

sample_path = "../data/cleaned/Ecommerce_cleaned_sample_100k.csv"
df_sample = pd.read_csv(sample_path)
print(f"Loaded clean sample dataset: {len(df_sample):,} rows")
df_sample.head()
```

## Step 2: Visualizations & Exploratory Data Analysis

Next, I'm computing batch-aggregated metrics across all 109.95M records to plot event type distributions, price log-histograms, daily activity trends, hourly UTC traffic, conversion funnels, and brand/category breakdowns.

```python
import matplotlib.pyplot as plt
import seaborn as sns

sns.set_theme(style="whitegrid", palette="deep")
event_counts = df_sample["event_type"].value_counts()

plt.figure(figsize=(8, 4))
sns.barplot(x=event_counts.index, y=event_counts.values, palette="viridis")
plt.title("Sample Event Type Distribution", weight="bold")
plt.ylabel("Count")
plt.show()
```
